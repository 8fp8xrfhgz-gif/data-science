### 2. (24%)

```python
# FIXED: the original version compared longitude and latitude
# separately using if/elif, so it never actually computed a real
# distance. i and h stayed at their initial value of 0, so the
# result printed as (0, 0) -- the nearest station was never found.
#
# Correct approach: for each station, compute the actual Euclidean
# distance using both longitude and latitude together, and keep
# track of the station with the smallest distance.

from math import sqrt

def EPAName1(X, Y, filename):
    X = float(X)
    Y = float(Y)
    fhand = open(filename, 'r')
    print('輸入的X,Y:(', X, ',', Y, ')')

    nearest_name = None
    nearest_lon = None
    nearest_lat = None
    min_distance = None

    for line in fhand:
        line = line.rstrip()
        fields = line.split(',')
        try:
            lon = float(fields[6])
            lat = float(fields[7])
        except ValueError:
            continue  # skip header row or malformed lines

        distance = sqrt((lon - X)**2 + (lat - Y)**2)

        if min_distance is None or distance < min_distance:
            min_distance = distance
            nearest_name = fields[0]
            nearest_lon = lon
            nearest_lat = lat

    fhand.close()
    print('最鄰近測站:', nearest_name)
    print('最鄰近測站的X,Y: (', nearest_lon, ',', nearest_lat, ')')
    print('距離:', min_distance)
    return [nearest_name, min_distance]
```

```python
EPAName1(X = 121.409738, Y = 25.183711, filename = 'EPA_STN1.csv')
```

### 3. (18%)

```python
# FIXED: the original function definition had no body (empty),
# so it couldn't have run on its own -- the output shown in the
# original submission relied on a leftover version of the function
# still in memory from an earlier, deleted cell. Re-running the
# notebook from scratch would have failed at this point.
#
# Logic: read the file, sort by the specified pollutant column T,
# and return the top n stations with the highest values.

POLLUTANT_COLUMN_INDEX = {
    'PSI': 10,
    'PM': 11,
    'O3': 12,
    'SO2': 13,
    'CO': 14,
    'NO2': 15,
}

def EPAName2(T, n, filename):
    col_idx = POLLUTANT_COLUMN_INDEX[T]
    fhand = open(filename, 'r')
    results = []

    for line in fhand:
        line = line.rstrip()
        fields = line.split(',')
        try:
            value = float(fields[col_idx])
        except (ValueError, IndexError):
            continue  # skip header row or malformed lines
        results.append((value, fields[0]))

    fhand.close()
    results.sort(reverse=True)  # descending by pollutant value
    return results[:n]
```

```python
EPAName2(T = "PM",  n = 4, filename = 'EPA_STN1.csv')
```
