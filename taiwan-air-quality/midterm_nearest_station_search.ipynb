### 1. (18%)

```python
fhand = open('EPA_STN1.csv', 'r')
x = input('請輸入測站站名(Name):')
d = dict()
for line in fhand:
    line = line.rstrip()
    if line.startswith(x):
        fields = line.split(',')
        PSI = fields[10]
        d['站名'] = x
        d['PSI數值'] = PSI
        print('dictionary格式查詢結果：' , d)
        if int(PSI) > 70 :
            print('空氣品質狀況: PSI數值>70，該地區空氣品質不佳')
            break
        elif int(PSI) <= 70 :
            print('空氣品質狀況: PSI數值<=70，該地區空氣品質良好')
            break
fhand.close()
```
